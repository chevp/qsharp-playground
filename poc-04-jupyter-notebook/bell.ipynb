# Bell-state experiment
Modern QDK in Jupyter: define a Q# operation with `%%qsharp`, run it from Python, plot the histogram.

```python
import qsharp
from qsharp_widgets import Histogram
```

```python
%%qsharp
import Std.Measurement.*;

operation Bell() : (Result, Result) {
    use (q0, q1) = (Qubit(), Qubit());
    H(q0);
    CNOT(q0, q1);
    return (MResetZ(q0), MResetZ(q1));
}
```

```python
results = qsharp.run("Bell()", shots=1000)
Histogram(results)
```
